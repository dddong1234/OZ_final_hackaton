# P1 dense enrichment LightGBM
입력은 P1 26점수 + row-local burden/type/topology뿐이다. raw 4천 유전자와 test는 사용하지 않는다. H1 단독 screen만 실행한다; blend H2는 게이트 통과 후에만 별도 생성한다.

```python
from pathlib import Path
import subprocess, sys
ROOT=next(p for p in (Path.cwd(),*Path.cwd().parents) if (p/'experiments/gs/notebooks/exp_model_002/common/run_p1_axis.py').exists())
RUNNER=ROOT/'experiments/gs/notebooks/exp_model_002/common/run_p1_axis.py'; RESULT=RUNNER.parent.parent/'result'
RUN_ID='exp-dense-enrichment-lgbm-01'; SEED=42; RUN_EXPERIMENT=True
# 고정: leaves=15, depth=4, lr=.03, estimators=500, min_child=30, reg_lambda=5, reg_alpha=.2
```

```python
if RUN_EXPERIMENT: subprocess.run([sys.executable,str(RUNNER),'--axis','dense','--seed',str(SEED),'--run-id',RUN_ID],check=True)
import pandas as pd, matplotlib.pyplot as plt
s=pd.read_csv(RESULT/f'{RUN_ID}_seed{SEED}_summary.csv'); assert s.leakage_check.all() and s.nan_as_mutation_count.eq(0).all(); display(s)
s.set_index('variant').oof_macro_f1.plot.bar(figsize=(7,4),ylim=(0.4,0.6),title='P1 vs dense LightGBM'); plt.tight_layout(); plt.show()
print('3-seed 확장 조건: P1 대비 +0.003 이상, 5 fold 중 4개 이상 상승, 안전성 통과.')
```
