# H1 — 자동 혼동 그룹 Mixture-of-Experts

H0의 26개 클래스 확률에서 자동으로 발견한 혼동 그룹 내부만 재분류합니다. 그룹 탐지·전처리·전문가 학습은 각 outer-fold train 안에서만 수행하며, seed42 screen에서는 test를 읽지 않습니다.

실행 순서: 1) 설정 셀, 2) 실행 셀, 3) 결과·판정 셀. 전체 실행은 약 1시간 이상 걸릴 수 있습니다.

```python
from pathlib import Path
import subprocess, sys
from tqdm.auto import tqdm

ROOT = Path('/Users/admin/Documents/FinalProject/OZ_fianl_hackaton')
RUNNER = ROOT / 'experiments/gs/notebooks/exp_model_006/common/run_h1_auto_confusion_moe.py'
RESULT = ROOT / 'experiments/gs/notebooks/exp_model_006/result'
RUN_ID = 'exp-h1-auto-confusion-moe-01'
RUN_EXPERIMENT = True
assert RUNNER.exists()
print({'runner': RUNNER, 'result': RESULT, 'seed': 42, 'test_read': False})
```

```python
if RUN_EXPERIMENT:
    command = [sys.executable, str(RUNNER), '--run-id', RUN_ID]
    process = subprocess.Popen(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    tail = []
    for line in tqdm(process.stdout, desc='H1 auto-confusion MoE', unit='line'):
        print(line, end='')
        tail = (tail + [line])[-120:]
    if process.wait():
        raise RuntimeError('H1 runner failed:\n' + ''.join(tail))
else:
    print('RUN_EXPERIMENT=False: 기존 결과만 읽습니다.')
```

```python
import json
import matplotlib.pyplot as plt
import pandas as pd

summary = pd.read_csv(RESULT / f'{RUN_ID}_seed42_summary.csv')
folds = pd.read_csv(RESULT / f'{RUN_ID}_seed42_fold_metrics.csv')
groups = pd.read_csv(RESULT / f'{RUN_ID}_seed42_auto_groups.csv')
classes = pd.read_csv(RESULT / f'{RUN_ID}_seed42_class_metrics.csv')
low_margin = pd.read_csv(RESULT / f'{RUN_ID}_seed42_low_margin.csv')
audit = json.loads((RESULT / f'{RUN_ID}_seed42_leakage_audit.json').read_text())
assert summary.leakage_check.all() and summary.nan_as_mutation_count.eq(0).all()
assert audit['test_read'] is False and audit['leakage_check'] is True
display(summary)
display(groups)
display(low_margin)
print('자동 판정:', audit['decision'])

folds.pivot(index='fold', columns='variant', values='macro_f1').plot(marker='o', figsize=(8, 4), title='H0 vs H1: fold Macro F1')
plt.tight_layout(); plt.show()
classes.sort_values('delta').set_index('class').delta.plot.barh(figsize=(7, 7), title='H1 class F1 delta vs H0')
plt.tight_layout(); plt.show()
```
