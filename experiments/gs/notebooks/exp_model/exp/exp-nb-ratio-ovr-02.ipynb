# exp-nb-ratio-ovr-02 — signature NB-ratio OVR 3-seed 확정 검증

## 고정 후보
- 입력: `event_core_signature`
- NB-ratio smoothing: `1.0`
- 최종 확률: `0.75 × 08-spec LR + 0.25 × NB-ratio OVR`
- CV seeds: `42 / 777 / 2024`, Stratified 5-fold

새 피처·smoothing·blend 비율은 탐색하지 않는다. 모든 feature 통계와 signature 선택은 outer fold-train에서만 계산하며 test는 읽지 않는다.


```python
from pathlib import Path
import subprocess, sys, pandas as pd, matplotlib.pyplot as plt
ROOT=next(p for p in (Path.cwd(),*Path.cwd().parents) if (p/'experiments/gs/notebooks/exp_model/common/exp_nb_ratio_ovr_runner.py').exists())
RUNNER=ROOT/'experiments/gs/notebooks/exp_model/common/exp_nb_ratio_ovr_runner.py'; RESULT=RUNNER.parent.parent/'result'
RUN_ID='exp-nb-ratio-ovr-02'; SEEDS=(42,777,2024); RUN_EXPERIMENT=True
if RUN_EXPERIMENT:
    for seed in SEEDS:
        print(f'\n===== seed {seed} =====')
        subprocess.run([sys.executable,str(RUNNER),'--seed',str(seed),'--run-id',RUN_ID],check=True)

all_rows=pd.concat([pd.read_csv(RESULT/f'{RUN_ID}_seed{seed}_summary.csv') for seed in SEEDS],ignore_index=True)
winner=all_rows.query("variant == 'event_core_signature' and model == 'lr075_nb025'").copy()
assert len(winner)==3 and winner.leakage_check.all() and winner.nan_as_mutation_count.eq(0).all()
display(winner[['seed','oof_macro_f1','delta_vs_variant_lr','feature_count_mean','convergence_warning_count','leakage_check','nan_as_mutation_count']])
display(winner[['oof_macro_f1','delta_vs_variant_lr']].agg(['mean','std']))
winner.plot(x='seed',y='delta_vs_variant_lr',marker='o',legend=False,figsize=(7,4),title='NB-ratio OVR confirmation: paired delta vs LR'); plt.axhline(0,color='black'); plt.tight_layout(); plt.show()
```
