```python
from pathlib import Path
import subprocess,sys,pandas as pd
ROOT=next(p for p in (Path.cwd(),*Path.cwd().parents) if (p/'experiments/gs/notebooks/exp_model/common/exp_profile_model_runner.py').exists())
RUNNER=ROOT/'experiments/gs/notebooks/exp_model/common/exp_profile_model_runner.py'; RESULT=RUNNER.parent.parent/'result'; RUN_ID='exp-profile-model-02'; SEEDS=(42,777,2024); RUN=True
if RUN:
    for seed in SEEDS: subprocess.run([sys.executable,str(RUNNER),'--seed',str(seed),'--run-id',RUN_ID],check=True)
parts=[pd.read_csv(RESULT/f'{RUN_ID}_seed{s}_summary.csv') for s in SEEDS]; s=pd.concat(parts); target=s[(s.model.eq('complement_nb'))&(s.blend_weight_profile.eq(.25))]; assert target.leakage_check.all() and target.nan_as_mutation_count.eq(0).all(); display(target); display(target[['blend_oof_macro_f1','delta_vs_lr08']].agg(['mean','std']))
```
